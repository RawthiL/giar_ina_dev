```python
import glob
import os
import shutil

def remove_suffix(path):
    base = os.path.basename(path)
    new_base = base[:base.rfind('_')] 
    return os.path.join(os.path.dirname(path), new_base)

images_paths = glob.glob(os.path.join('/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/nuevos datasets/nuevos datasets', '*/*/*'))
images_paths = images_paths + glob.glob(os.path.join('/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/images/ina/images', '*'))
images_paths = images_paths + glob.glob(os.path.join('/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/images/onion_cell_merged/images', '*/*'))

anaphase_crops = glob.glob(os.path.join('/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/cropped_images/classification/test_v2/anaphase', '*'))
telophase_crops = (glob.glob(os.path.join('/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/cropped_images/classification/test_v2/telophase', '*')))
crops = anaphase_crops + telophase_crops

crops = [remove_suffix(p) for p in crops]
images_with_crops = [os.path.basename(p) for p in crops]

matched_images = []
for crop_name in images_with_crops:
    for img_path in images_paths:
        if crop_name in img_path:
            matched_images.append(img_path)
            break  # If you only want the first match, otherwise remove this line

matched_images = list(set(matched_images))

destination_folder = "/home/nicolas/Documentos/UTN/INA/giar_ina_dev/media/images/images_with_separetead_nucleai"
os.makedirs(destination_folder, exist_ok=True)

for img_path in matched_images:
    if os.path.isfile(img_path):
        shutil.copy(img_path, destination_folder)
```
