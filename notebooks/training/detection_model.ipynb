```python
import os
import sys

from IPython.display import display
from PIL import Image as PILImage
from ultralytics import YOLO
```

```python
sys.path.append("../../")
os.chdir("../../")
```

```python
model = YOLO("src/allium_cepa_classifier/models/weights/yolo11n.pt")
```

```python
results = model.train(
    data="datasets/yolo_dataset/data.yaml", epochs=200, imgsz=640, device=0, name="train"
)
```

```python
metrics = model.val(
    split="val",  # "val" (default) or "test"
    imgsz=640,
    batch=16,
    conf=0.001,  # score threshold for eval curves
    iou=0.7,  # NMS IoU for mAP calc
    plots=True,  # save PR curves, confusion matrix, etc.
)
print(metrics)
```

```python
model.save("src/allium_cepa_classifier/models/weights/object_detection.pt")
```
